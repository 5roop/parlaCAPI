#  Get one sample datum from the /sample endpoint

````python
url = "https://parlacap.ipipan.waw.pl/api/v1/"

import requests
from dotenv import load_dotenv
import os

load_dotenv(dotenv_path="my_secrets.env")
headers = {"X-API-Key": os.getenv("parlacapi_key")}
response = requests.get(url + "sample?size=1", headers=headers)
if not response.status_code == 200:
    raise Exception(
        f"Got weird response code: {response.status_code}, content: {response.content}"
    )

payload = response.json()
payload
````

````python
filter = {
    "filter": {
        "operator": "AND",
        "filters": [
            {"column": "text", "operator": "CONTAINS", "value": "krava"},
            {
                "operator": "OR",
                "filters": [
                    {"column": "parliament", "operator": "=", "value": "RS"},
                    {"column": "parliament", "operator": "=", "value": "SI"},
                ],
            },
        ],
    },
    # "search": {"query": "krava", "field": "text"},
    "columns": ["speaker_name", "date", "text", "cap_category"],
    "limit": 5,
    "offset": 0,
}

url = "https://parlacap.ipipan.waw.pl/api/v1/"

import requests
from dotenv import load_dotenv
import os

load_dotenv(dotenv_path="my_secrets.env")
headers = {"X-API-Key": os.getenv("parlacapi_key")}

response = requests.post(url + "filter", json=filter, headers=headers)
if not response.status_code == 200:
    raise Exception(
        f"Got weird response code: {response.status_code}, response content: {response.content}"
    )

payload = response.json()
payload
````

# Downloading


````python
filter = {
    "filter": {
        "operator": "AND",
        "filters": [
            {"column": "text", "operator": "CONTAINS", "value": "krava"},
            {
                "operator": "OR",
                "filters": [
                    {"column": "parliament", "operator": "=", "value": "RS"},
                    {"column": "parliament", "operator": "=", "value": "SI"},
                ],
            },
        ],
    },
    # "search": {"query": "krava", "field": "text"},
    "columns": ["speaker_name", "date", "text", "cap_category"],
    "limit": 100,
    "offset": 0,
}

url = "https://parlacap.ipipan.waw.pl/api/v1/"

import requests
from dotenv import load_dotenv
import os

load_dotenv(dotenv_path="my_secrets.env")
headers = {"X-API-Key": os.getenv("parlacapi_key")}
response = requests.post(url + "download?format=csv", json=filter, headers=headers)
if not response.status_code == 200:
    raise Exception(
        f"Got weird response code: {response.status_code}, response content: {response.content}"
    )

from pathlib import Path

Path("my_downloaded_file.csv").write_bytes(response.content)
````
